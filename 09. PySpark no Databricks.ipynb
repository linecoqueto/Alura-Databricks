```python
data = [
    {"Filme A", "Drama", 120},
    {"Filme B", "Ação", 95},
    {"Filme C", "Comédia", 110}
]

df = spark.createDataFrame(
    data,
    ["titulo","categoria","duração"]
)

display(df)
```

```python
from pyspark.sql.functions import upper

df_transformado = df.withColumn(
    "catogoria_mauscula",
    upper("categoria")
)

display(df_transformado)
```

```python
#Criando um view temporária para consultas SQL
df_transformado.createOrReplaceTempView("filmes")
```

```python
%sql
Select * from filmes
limit 5
```

```python
%sql
--Consultando com SQL e calculando a média

SELECT
    categoria,
    AVG(duracao) as media_duracao
FROM filmes
GROUP BY categoria
```

```python
#Reproduzindo a agregação com spark

from pyspark.sql.functions import avg

df.groupBy('categoria')\
  .agg(avg('duracao').alias('media_duracao'))\
  .display()
```

### *Exemplo utilizando uma transformação*

```python
data = [("Filme A", "Drama", 120),
        ("Filme B", "Ação", 95),
        ("Filme C", "Drama", 110),
        ("Filme D", "Comédia", 100)]

df = spark.createDataFrame(data)

df = df.toDF('titulo', 'categoria', 'duracao')
```

```python
display(df)
```

```python
df_filtrado = df.filter(df.duracao > 100)
```

```python
display(df_filtrado)
```
