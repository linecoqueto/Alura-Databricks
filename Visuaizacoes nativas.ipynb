```python
%sql
select
count(*) as total_historico,
count(data_desligamento) as total_desligamento,
round(count(data_desligamento) * 100.0 / count(*),2) as turnover_percentual
from workspace.default.funcionarios;
```

```python
%sql
select departamento,
  count(*) as total_funcionarios,
  count(data_desligamento) as total_desligamento,
  round(count(data_desligamento) * 100.0 / count(*), 2) as turnover_percentual
from workspace.default.funcionarios
group by departamento
order by turnover_percentual desc
```
